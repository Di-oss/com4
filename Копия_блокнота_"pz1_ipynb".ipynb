```r

# Установка и загрузка пакетов
install.packages("DAAG")
install.packages("ggplot2")
install.packages("psych")
install.packages("car")
install.packages("hexbin")

library(DAAG)
library(ggplot2)
library(psych)
library(car)
library(hexbin)

# Загрузка данных cars
data(cars)
cars
describe(cars)

# Диаграмма рассеяния
ggplot(cars, aes(x = speed, y = dist)) +
  geom_point() +
  stat_smooth(method = "lm") +
  xlab("Скорость, миль/ч") +
  ylab("Тормозной путь, футов")

# Определение переменных и расчёт прогнозных значений
x <- cars$speed
y <- cars$dist

n <- dim(as.matrix(x))[1]; m <- dim(as.matrix(x))[2]
M_reg <- lm(y ~ x); pred <- predict(M_reg)

# Оценка качества линейной модели (вручную)
RSS <- sum((y - pred) * (y - pred))
RMSE <- sqrt(RSS / n)
RSE <- sqrt(RSS / (n - m - 1))
c(RSS, RMSE, RSE)

Rsquared <- 1 - RSS / sum((mean(y) - y)^2)
Fcr <- (sum((mean(y) - pred)^2) / m) / (RSS / (n - m - 1))
p <- pf(q = Fcr, df1 = m, df2 = (n - m - 1), lower.tail = FALSE)
c(Rsquared, Fcr, p)

# Оценка качества модели (встроенные функции)
anova(M_reg)
summary(M_reg)

# Тест на гомоскедастичность
ncvTest(M_reg)

# Построение генеральной линейной модели (GLM)
M_glm <- glm(y ~ x)

# Оценка качества ГЛМ (вручную)
lgLik <- logLik(M_glm)
D.null <- M_glm$null.deviance
D <- M_glm$deviance
df <- with(M_glm, df.null - df.residual)
p <- pchisq(D.null - D, df, lower.tail = FALSE)
PRsquare <- 1 - D / D.null
c(lgLik, D, D.null, p, PRsquare)

# Оценка качества модели (функция)
summary(M_glm)
anova(M_glm, test = "Chisq")

# График с гексагонами
ggplot(cars, aes(x = speed, y = dist)) +
  geom_hex(binwidth = c(3, 20)) +   # подбираем под масштаб
  geom_smooth(method = "lm", color = "red", se = FALSE) +
  xlab("Скорость, миль/ч") +
  ylab("Тормозной путь, футов")

# Оценка «сложности» полиномов (вручную)
k <- extractAIC(M_glm)[1]
AIC_val <- AIC(M_glm)
AICc <- AIC(M_glm) + 2 * k * (k + 1) / (n - k - 1)
BIC_val <- BIC(M_glm)
c(AIC_val, AICc, BIC_val)

# Построение моделей со степенью полинома от 1 до 7
max.poly <- 7
AIC.BIC <- data.frame(
  criterion = c(rep("AIC", max.poly), rep("BIC", max.poly)),
  value = numeric(max.poly * 2),
  degree = rep(1:max.poly, times = 2)
)

for (i in 1:max.poly) {
  AIC.BIC[i, 2] <- AIC(lm(y ~ poly(x, i)))
  AIC.BIC[i + max.poly, 2] <- BIC(lm(y ~ poly(x, i)))
}

# График AIC и BIC для разных степеней полинома
qplot(degree, value, data = AIC.BIC,
      geom = "line", linetype = criterion) +
  xlab("Степень полинома") +
  ylab("Значение критерия")
```
