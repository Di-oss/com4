```r

# Установка и загрузка пакетов
install.packages("DAAG")
install.packages("ggplot2")
install.packages("psych")
install.packages("car")
install.packages("hexbin")

library(DAAG)
library(ggplot2)
library(psych)
library(car)
library(hexbin)

# Загрузка данных Titanic
#    Используем встроенный набор titanic_train
install.packages("titanic")
library(titanic)
data("titanic_train")

titanic <- titanic_train

# Предварительная обработка: удаляем строки с NA в Age и Survived,
# а также приводим Survived к числовому виду для y
titanic <- titanic[!is.na(titanic$Age), ]          # удаляем пропуски в возрасте
titanic$Survived_num <- as.numeric(titanic$Survived) - 1   # 0/1

# Просмотр данных
head(titanic)

# Описательная статистика
describe(titanic[, c("Age", "Survived_num", "Pclass", "Sex")])

# Диаграмма рассеяния
ggplot(titanic, aes(x = Age, y = Survived_num)) +
  geom_point() +
  stat_smooth(method = "lm") +
  xlab("Возраст, лет") +
  ylab("Выживание (0/1)")

# Определение переменных и расчёт прогнозных значений
x <- titanic$Age
y <- titanic$Survived_num   # аналог ohms/1000

n <- dim(as.matrix(x))[1]
m <- dim(as.matrix(x))[2]

M_reg <- lm(y ~ x)
pred <- predict(M_reg)

# Оценка качества линейной модели (вручную)
RSS <- sum((y - pred) * (y - pred))
RMSE <- sqrt(RSS / n)
RSE <- sqrt(RSS / (n - m - 1))
c(RSS, RMSE, RSE)

Rsquared <- 1 - RSS / sum((mean(y) - y)^2)
Fcr <- (sum((mean(y) - pred)^2) / m) / (RSS / (n - m - 1))
p <- pf(q = Fcr, df1 = m, df2 = (n - m - 1), lower.tail = FALSE)
c(Rsquared, Fcr, p)

# Оценка качества линейной модели
anova(M_reg)
summary(M_reg)


install.packages("car")
library(car)
ncvTest(M_reg)

# Построение генеральной линейной модели (GLM)
M_glm <- glm(y ~ x, family = binomial)

# Оценка качества GLM (вручную, как в исходном ПЗ)
lgLik <- logLik(M_glm)
D.null <- M_glm$null.deviance
D <- M_glm$deviance
df <- with(M_glm, df.null - df.residual)
p <- pchisq(D.null - D, df, lower.tail = FALSE)
PRsquare <- 1 - D / D.null
c(lgLik, D, D.null, p, PRsquare)

# Оценка качества GLM (функции summary, anova)
summary(M_glm)
anova(M_glm, test = "Chisq")

# График с гексагонами для определения характера модели
ggplot(titanic, aes(x = Age, y = Survived_num)) +
  geom_hex(binwidth = c(3, 0.1)) +
  geom_smooth(method = "glm", method.args = list(family = "binomial"),
              color = "red", se = FALSE) +
  xlab("Возраст, лет") +
  ylab("Выживание (0/1)")

# Оценка «сложности» полиномов (вручную, как в исходном)
k <- extractAIC(M_glm)[1]
AIC_val <- extractAIC(M_glm)[2]
AIC_val <- AIC(M_glm)
AICc <- AIC(M_glm) + 2 * k * (k + 1) / (n - k - 1)
BIC_val <- BIC(M_glm)
BIC_val <- AIC(M_glm, k = log(n))
c(AIC_val, AICc, BIC_val)

# Построение моделей со степенью полинома от 1 до 7
max.poly <- 7
AIC.BIC <- data.frame(criterion = c(rep("AIC", max.poly),
                                    rep("BIC", max.poly)),
                      value = numeric(max.poly * 2),
                      degree = rep(1:max.poly, times = 2))

for (i in 1:max.poly) {
  # Для полиномиальной регрессии используем lm (как в исходном ПЗ)
  AIC.BIC[i, 2] <- AIC(lm(y ~ poly(x, i)))
  AIC.BIC[i + max.poly, 2] <- BIC(lm(y ~ poly(x, i)))
}

# График AIC и BIC для разных степеней полинома
qplot(degree, value, data = AIC.BIC,
      geom = "line", linetype = criterion) +
  xlab("Степень полинома") +
  ylab("Значение критерия")
```
