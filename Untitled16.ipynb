```r

# 1. Установка и загрузка необходимых пакетов

install.packages("ggplot2")
install.packages("hexbin")
install.packages("psych")

library(ggplot2)
library(hexbin)
library(psych)

# Пакеты для работы с данными Titanic и загрузки из Kaggle
install.packages("titanic")  # Содержит данные Titanic
install.packages("RKaggle")  # Для загрузки данных с Kaggle

library(titanic)
library(RKaggle)


# 2. Загрузка и подготовка данных Titanic
data("titanic_train")

# Изучаем структуру данных
str(titanic_train)

# Создаем копию набора данных для работы
titanic <- titanic_train

# Приводим данные к формату, пригодному для анализа:
# - Преобразуем категориальные переменные в факторы
titanic$Survived <- as.factor(titanic$Survived)
titanic$Pclass <- as.factor(titanic$Pclass)
titanic$Sex <- as.factor(titanic$Sex)
titanic$Embarked <- as.factor(titanic$Embarked)

# - Удаляем столбцы, которые не планируем использовать в модели
titanic <- subset(titanic, select = -c(PassengerId, Name, Ticket, Cabin))

# - Удаляем строки с пропущенными значениями (для простоты)
titanic <- na.omit(titanic)

# Выводим статистическое описание данных
describe(titanic)

# 3. Визуализация данных и определение характера модели
# Поскольку целевая переменная 'Survived' бинарна (0/1), выбираем логистическую регрессию

# График зависимости выживания от возраста
ggplot(titanic, aes(x = Age, y = as.numeric(Survived) - 1)) +
  geom_point(alpha = 0.5) +
  geom_smooth(method = "glm", method.args = list(family = "binomial"), se = FALSE, color = "red") +
  xlab("Возраст, лет") +
  ylab("Вероятность выживания") +
  ggtitle("Зависимость вероятности выживания от возраста")

# График с гексагонами для оценки плотности распределения
ggplot(titanic, aes(x = Age, y = as.numeric(Survived) - 1)) +
  geom_hex(bins = 30) +
  geom_smooth(method = "glm", method.args = list(family = "binomial"), se = FALSE, color = "red") +
  xlab("Возраст, лет") +
  ylab("Вероятность выживания") +
  ggtitle("Плотность распределения возраста и выживания")

# 4. Построение и оценка качества линейной модели

# Определяем переменные
x <- titanic$Age
y <- as.numeric(titanic$Survived) - 1  # Преобразуем в числовую (0/1)

# Расчет прогнозных значений
n <- dim(as.matrix(x))[1]
m <- dim(as.matrix(x))[2]
M_reg <- lm(y ~ x)
pred <- predict(M_reg)

# Оценка качества модели (вручную)
RSS <- sum((y - pred) * (y - pred))
RMSE <- sqrt(RSS / n)
RSE <- sqrt(RSS / (n - m - 1))
cat("RSS:", RSS, "RMSE:", RMSE, "RSE:", RSE, "\n")

Rsquared <- 1 - RSS / sum((mean(y) - y)^2)
Fcr <- (sum((mean(y) - pred)^2) / m) / (RSS / (n - m - 1))
p <- pf(q = Fcr, df1 = m, df2 = (n - m - 1), lower.tail = FALSE)
cat("R-squared:", Rsquared, "F-statistic:", Fcr, "p-value:", p, "\n")

# Оценка качества модели (встроенная функция)
anova(M_reg)
summary(M_reg)

# 5. Построение и оценка качества генеральной линейной модели (логистической регрессии)
# Для бинарного отклика используем семейство binomial с логистической функцией связи (logit).

M_glm <- glm(Survived ~ Age, data = titanic, family = binomial(link = "logit"))

# Оценка качества ГЛМ (вручную)
lgLik <- logLik(M_glm)
D.null <- M_glm$null.deviance
D <- M_glm$deviance
df <- with(M_glm, df.null - df.residual)
p <- pchisq(D.null - D, df, lower.tail = FALSE)
PRsquare <- 1 - D / D.null
cat("Log-Likelihood:", lgLik, "Deviance:", D, "Null Deviance:", D.null, "p-value:", p, "Pseudo R-squared:", PRsquare, "\n")

# Оценка качества модели (функция)
summary(M_glm)
anova(M_glm, test = "Chisq")  # Используем хи-квадрат тест для сравнения моделей

# 6. Оценка «сложности» полиномов с использованием AIC, AICc и BIC
# Проверяем, не улучшится ли модель при добавлении полиномиальных членов возраста.

max.poly <- 7

# Создание пустой таблицы для хранения значений AIC и BIC
AIC.BIC <- data.frame(
  criterion = c(rep("AIC", max.poly), rep("BIC", max.poly)),
  value = numeric(max.poly * 2),
  degree = rep(1:max.poly, times = 2)
)

for (i in 1:max.poly) {
  # Для полиномиальной регрессии используем lm, но оценка качества через AIC/BIC
  poly_model <- glm(Survived ~ poly(Age, i), data = titanic, family = binomial(link = "logit"))
  AIC.BIC[i, 2] <- AIC(poly_model)
  AIC.BIC[i + max.poly, 2] <- BIC(poly_model)
}

# График AIC и BIC для разных степеней полинома
ggplot(AIC.BIC, aes(x = degree, y = value, linetype = criterion)) +
  geom_line() +
  xlab("Степень полинома") +
  ylab("Значение критерия") +
  ggtitle("Сравнение AIC и BIC для полиномиальных моделей")
```
