```r
# Загрузка пакетов
library(data.table)
library(cluster)
library(reshape2)
library(ggplot2)

# Путь к CSV (из вывода Python-ячейки)
csv_path <- "/kaggle/input/world-development-indicators/wdi-csv-zip-57-mb-/WDIData.csv"
if (!file.exists(csv_path)) stop("Файл не найден")

# Чтение данных с заголовками
data_wide <- fread(csv_path, header = TRUE)

# Список желаемых индикаторов и их возможных точных названий в данных
# Используем список с альтернативными названиями для лучшего сопоставления
indicator_map <- list(
  "GDP per capita" = c("GDP per capita (current US$)"),
  "Population" = c("Population, total"),
  "Health expenditure per capita" = c("Health expenditure per capita (current US$", "Current health expenditure per capita (current US$)"),
  "School enrollment, primary" = c("School enrollment, primary (% gross)", "School enrollment, primary, gross %"),
  "CO2 emissions per capita" = c("CO2 emissions (metric tons per capita)")
)

found_indicators_for_analysis <- c()
all_available_names <- unique(data_wide$`Indicator Name`)

# Ищем лучшие совпадения для каждого индикатора
for (category in names(indicator_map)) {
  matched_name <- NULL
  for (potential_name in indicator_map[[category]]) {
    if (potential_name %in% all_available_names) {
      matched_name <- potential_name
      break # Найдено совпадение для этой категории
    }
  }
  if (!is.null(matched_name)) {
    found_indicators_for_analysis <- c(found_indicators_for_analysis, matched_name)
  } else {
    warning(paste0("Не удалось найти подходящий индикатор для категории '", category, "'. Он будет пропущен."))
  }
}

# Обновляем список индикаторов, используя найденные названия
indicators <- unique(found_indicators_for_analysis)

if (length(indicators) < 2) {
  stop("Слишком мало индикаторов найдено в данных для анализа. Проверьте ключевые слова индикаторов или источник данных.")
}

cat("Окончательный список индикаторов, выбранных для анализа:\n")
print(indicators)

# Выбираем год
target_year <- "2017"

# Проверяем, есть ли нужный год в данных
years <- names(data_wide)[!names(data_wide) %in% c("Country Name", "Country Code", "Indicator Name", "Indicator Code")]
if (!target_year %in% years) {
  stop(paste("Год", target_year, "не найден. Доступные годы:", paste(head(years, 10), collapse = ", ")))
}

# Фильтруем строки по найденным индикаторам и оставляем нужные колонки
data_filtered <- data_wide[`Indicator Name` %in% indicators, c("Country Name", "Indicator Name", target_year), with = FALSE]

# Удаляем строки, где значение для целевого года равно NA
data_filtered <- na.omit(data_filtered)

# Проверяем, какие индикаторы остались после удаления NA
valid_indicators_after_na_filter <- unique(data_filtered$`Indicator Name`)
if (length(setdiff(indicators, valid_indicators_after_na_filter)) > 0) {
  warning("Некоторые индикаторы не имели действительных данных для целевого года ни для одной страны и были исключены: ", paste(setdiff(indicators, valid_indicators_after_na_filter), collapse = ", "))
  indicators <- valid_indicators_after_na_filter # Обновляем список индикаторов
}

if (length(indicators) < 2) {
  stop("После фильтрации осталось слишком мало индикаторов с полными данными для целевого года. Измените индикаторы или год.")
}

# Переименовываем колонку с годом в "Value" для dcast
setnames(data_filtered, target_year, "Value")

# Разворачиваем в широкий формат: строки – страны, колонки – индикаторы
data_indicators_wide <- dcast(data_filtered, `Country Name` ~ `Indicator Name`, value.var = "Value")

# Удаляем страны, у которых есть пропуски хотя бы по одному из индикаторов (после dcast)
original_country_count <- nrow(data_indicators_wide)
data_indicators_wide_clean <- na.omit(data_indicators_wide)

if (nrow(data_indicators_wide_clean) < original_country_count) {
  cat(paste0("Удалено ", original_country_count - nrow(data_indicators_wide_clean), " стран из-за отсутствия данных по некоторым индикаторам.\n"))
}

# Проверяем, достаточно ли стран осталось для кластеризации
if (nrow(data_indicators_wide_clean) < 3) {
  stop("Слишком мало стран с полными данными по всем выбранным индикаторам для кластеризации. Попробуйте другой год или измените список индикаторов.")
}

# Устанавливаем названия строк и удаляем колонку со странами
rownames(data_indicators_wide_clean) <- data_indicators_wide_clean$`Country Name`
my_data <- data_indicators_wide_clean[, -1, drop = FALSE]

cat("Размер матрицы после обработки:", dim(my_data), "\n")

# ------------------- КЛАСТЕРИЗАЦИЯ -------------------
# Тепловая карта (формируем «дендрограмму»)
x <- as.matrix(my_data)
rc <- rainbow(nrow(x), start = 0, end = .3)
cc <- rainbow(ncol(x), start = 0, end = .3)
heatmap(x, scale = "col", RowSideColors = rc, ColSideColors = cc,
        margins = c(10,10), cexCol = 1.5, cexRow = 1)

# Оптимальное число кластеров
df.stand <- scale(my_data)
wss <- sapply(1:10, function(k) kmeans(df.stand, k, nstart = 10)$tot.withinss)
plot(1:10, wss, type = "b", pch = 19, xlab = "Число кластеров K",
     ylab = "Общая внутригрупповая сумма квадратов")

# K-means с 4 кластерами (можно изменить)
set.seed(123)
clus <- kmeans(df.stand, centers = 4)
print(clus)

# Таблица расстояний между кластерами
n <- nrow(df.stand)
euc.dist <- as.matrix(dist(df.stand))
dist_melt <- melt(euc.dist)
df.stand <- as.data.frame(df.stand)
df.stand$cluster <- clus$cluster
pairs <- data.frame(dist = dist_melt,
                    ca = as.vector(outer(1:n, 1:n, function(a,b) df.stand$cluster[a])),
                    cb = as.vector(outer(1:n, 1:n, function(a,b) df.stand$cluster[b])))
dist_between <- dcast(pairs, ca ~ cb, value.var = "dist.value", mean)
print(dist_between)

# PCA и визуализация
pca <- prcomp(my_data, center = TRUE, scale = TRUE)
d <- data.frame(x = pca$x[,1], y = pca$x[,2], cluster = clus$cluster)

h <- do.call(rbind, lapply(unique(d$cluster), function(c) {
  f <- subset(d, cluster == c)
  f[chull(f), ]
}))

ggplot(d, aes(x, y, color = as.factor(cluster), label = cluster)) +
  geom_text(size = 3) +
  geom_polygon(data = h, aes(group = cluster), alpha = 0.4, linetype = 0) +
  theme(legend.position = "none") +
  labs(title = "Кластеризация стран по экономическим показателям")

```

```r
# ======================================================================
# Второе задание: анализ структуры данных (PCA, RDA, MANOVA, Hotelling)
# на основе кластеров, полученных по индикаторам World Development Indicators
# ======================================================================

# 1. Установка и загрузка пакетов ----------------------------------------
if (!require(data.table)) install.packages("data.table")
if (!require(cluster)) install.packages("cluster")
if (!require(reshape2)) install.packages("reshape2")
if (!require(ggplot2)) install.packages("ggplot2")
if (!require(vegan)) install.packages("vegan")
if (!require(Hotelling)) install.packages("Hotelling")

library(data.table)
library(cluster)
library(reshape2)
library(ggplot2)
library(vegan)
library(Hotelling)

# 2. Загрузка и подготовка данных ----------------------------------------
# Путь к CSV (используйте тот же, что и в первом задании)
csv_path <- "/kaggle/input/world-development-indicators/wdi-csv-zip-57-mb-/WDIData.csv"
if (!file.exists(csv_path)) stop("Файл не найден. Проверьте путь.")

data_wide <- fread(csv_path, header = TRUE)

# Список желаемых индикаторов (можно изменить)
indicator_map <- list(
  "GDP per capita" = c("GDP per capita (current US$)"),
  "Population" = c("Population, total"),
  "Health expenditure per capita" = c("Health expenditure per capita (current US$", "Current health expenditure per capita (current US$)"),
  "School enrollment, primary" = c("School enrollment, primary (% gross)", "School enrollment, primary, gross %"),
  "CO2 emissions per capita" = c("CO2 emissions (metric tons per capita)")
)

# Поиск доступных индикаторов
found_indicators <- c()
all_names <- unique(data_wide$`Indicator Name`)
for (category in names(indicator_map)) {
  matched <- NULL
  for (name in indicator_map[[category]]) {
    if (name %in% all_names) {
      matched <- name
      break
    }
  }
  if (!is.null(matched)) {
    found_indicators <- c(found_indicators, matched)
  } else {
    warning(paste("Не найден индикатор для категории", category))
  }
}
if (length(found_indicators) < 2) stop("Найдено слишком мало индикаторов")

cat("Выбранные индикаторы:\n")
print(found_indicators)

# Год анализа (доступны до 2018)
target_year <- "2017"   # при необходимости измените на "2018", "2016" и т.д.
years <- names(data_wide)[!names(data_wide) %in% c("Country Name", "Country Code", "Indicator Name", "Indicator Code")]
if (!target_year %in% years) {
  stop(paste("Год", target_year, "не найден. Доступны:", paste(head(years, 10), collapse = ", ")))
}

# Фильтрация и подготовка матрицы "страны × индикаторы"
data_filtered <- data_wide[`Indicator Name` %in% found_indicators,
                           c("Country Name", "Indicator Name", target_year), with = FALSE]
setnames(data_filtered, target_year, "Value")
data_filtered <- na.omit(data_filtered)  # удаляем строки с NA

# Преобразование в широкий формат
data_wide_format <- dcast(data_filtered, `Country Name` ~ `Indicator Name`, value.var = "Value")
rownames(data_wide_format) <- data_wide_format$`Country Name`
my_data <- data_wide_format[, -1, drop = FALSE]
my_data <- na.omit(my_data)   # удаляем страны с пропусками

if (nrow(my_data) < 3) stop("Слишком мало стран для анализа. Измените год или индикаторы.")
cat("Размер матрицы:", dim(my_data), "\n")

# 3. Кластеризация (k-means, 4 кластера) ---------------------------------
set.seed(123)
df.stand <- scale(my_data)
wss <- sapply(1:10, function(k) kmeans(df.stand, k, nstart = 10)$tot.withinss)
# plot(1:10, wss, type = "b")  # для выбора числа кластеров – не обязателен
clus <- kmeans(df.stand, centers = 4)

# Присваиваем кластеры
clusters <- clus$cluster
cat("Размеры кластеров:\n")
print(table(clusters))

# 4. Базовые статистики по признакам (как в примере с Glass) -------------
stats <- t(apply(my_data, 2, function(x) {
  c(Минимум = min(x),
    Максимум = max(x),
    Среднее = mean(x),
    Отклонение = sd(x),
    Корреляция = cor(x, clusters))   # корреляция с номером кластера
}))
print(round(stats, 3))

# 5. Поиск структуры вслепую (PCA с помощью vegan::rda) ------------------
Y <- as.data.frame(my_data)
mod.pca <- rda(Y ~ 1)
summary(mod.pca, scale = TRUE, scaling = 2)

# Извлекаем координаты стран в пространстве первых двух PC
scores_pca <- scores(mod.pca, scaling = 2)
pca.scores <- as.data.frame(scores_pca$sites[, 1:2])
pca.scores$cluster <- as.factor(clusters)

# Выпуклые оболочки для каждого кластера
hull_list <- lapply(unique(pca.scores$cluster), function(c) {
  f <- subset(pca.scores, cluster == c)
  if (nrow(f) >= 3) f[chull(f[, 1:2]), ] else NULL
})
hull <- do.call(rbind, hull_list)

# Доли объясненной дисперсии
axX <- paste("PC1 (", as.integer(100 * mod.pca$CA$eig[1] / sum(mod.pca$CA$eig)), "%)")
axY <- paste("PC2 (", as.integer(100 * mod.pca$CA$eig[2] / sum(mod.pca$CA$eig)), "%)")

# График PCA с кластерами
ggplot() +
  geom_polygon(data = hull,
               aes(x = PC1, y = PC2, fill = cluster), alpha = 0.4, linetype = 0) +
  geom_point(data = pca.scores,
             aes(x = PC1, y = PC2, shape = cluster, colour = cluster), size = 3) +
  scale_colour_manual(values = rainbow(length(unique(clusters)))) +
  xlab(axX) + ylab(axY) + coord_equal() + theme_bw()

# 6. Поиск структуры с учетом известных кластеров (RDA) ------------------
FAC <- as.factor(clusters)
mod.rda <- rda(Y ~ FAC)
summary(mod.rda)

# Координаты сайтов в RDA-пространстве
rda.scores <- as.data.frame(scores(mod.rda, display = "sites", scaling = 3))
rda.scores <- cbind(FAC, rda.scores)

# Центроиды групп
centroids <- aggregate(cbind(RDA1, PC1) ~ FAC, rda.scores, mean)

# Стандартные ошибки
f_se <- function(z) sd(z) / sqrt(length(z))
se <- aggregate(cbind(RDA1, PC1) ~ FAC, rda.scores, f_se)
names(se) <- c("FAC", "RDA1.se", "PC1.se")
centroids <- merge(centroids, se, by = "FAC")

# График RDA с эллипсами доверия (уровень 0.8)
ggplot() +
  geom_point(data = rda.scores,
             aes(x = RDA1, y = PC1, shape = FAC, colour = FAC), size = 2) +
  stat_ellipse(data = rda.scores, aes(x = RDA1, y = PC1, fill = FAC),
               geom = "polygon", level = 0.8, alpha = 0.2) +
  geom_point(data = centroids, aes(x = RDA1, y = PC1, colour = FAC),
             shape = 1, size = 4) +
  scale_colour_manual(values = rainbow(length(unique(FAC)))) +
  scale_fill_manual(values = rainbow(length(unique(FAC)))) +
  theme_bw()

# 7. Оценка достоверности различий между кластерами ----------------------
# MANOVA для всех кластеров
mod.man <- manova(as.matrix(Y) ~ FAC)
cat("\nРезультаты MANOVA:\n")
print(summary(mod.man))

# Hotelling test для двух наиболее многочисленных кластеров
tab <- table(FAC)
largest <- names(sort(tab, decreasing = TRUE)[1:2])
group1 <- Y[FAC == largest[1], ]
group2 <- Y[FAC == largest[2], ]

cat("\nHotelling test для кластеров", largest[1], "и", largest[2], ":\n")
hot <- hotelling.test(group1, group2, perm = TRUE, B = 1000, progBar = FALSE)
print(hot)
```

```r

# 1. Установка и загрузка пакетов ----------------------------------------
if (!require(data.table)) install.packages("data.table")
if (!require(cluster)) install.packages("cluster")
if (!require(reshape2)) install.packages("reshape2")
if (!require(ggplot2)) install.packages("ggplot2")
if (!require(vegan)) install.packages("vegan")
if (!require(Hotelling)) install.packages("Hotelling")

library(data.table)
library(cluster)
library(reshape2)
library(ggplot2)
library(vegan)
library(Hotelling)

# 2. Загрузка и подготовка данных ----------------------------------------
csv_path <- readLines("/tmp/csv_path.txt", warn = FALSE)[1]
if (is.na(csv_path) || !file.exists(csv_path)) {
  stop("Файл не найден. Убедитесь, что Python-ячейка выполнена и путь сохранён.")
}
data_wide <- fread(csv_path, header = TRUE)

# Список желаемых индикаторов
indicator_map <- list(
  "GDP per capita" = c("GDP per capita (current US$)"),
  "Population" = c("Population, total"),
  "Health expenditure per capita" = c("Health expenditure per capita (current US$", "Current health expenditure per capita (current US$)"),
  "School enrollment, primary" = c("School enrollment, primary (% gross)", "School enrollment, primary, gross %"),
  "CO2 emissions per capita" = c("CO2 emissions (metric tons per capita)")
)

# Поиск доступных индикаторов
found_indicators <- c()
all_names <- unique(data_wide$`Indicator Name`)
for (category in names(indicator_map)) {
  matched <- NULL
  for (name in indicator_map[[category]]) {
    if (name %in% all_names) {
      matched <- name
      break
    }
  }
  if (!is.null(matched)) {
    found_indicators <- c(found_indicators, matched)
  } else {
    warning(paste("Не найден индикатор для категории", category))
  }
}
if (length(found_indicators) < 2) stop("Найдено слишком мало индикаторов")
cat("Выбранные индикаторы:\n")
print(found_indicators)

# Год анализа (доступны до 2018)
target_year <- "2015"
years <- names(data_wide)[!names(data_wide) %in% c("Country Name", "Country Code", "Indicator Name", "Indicator Code")]
if (!target_year %in% years) {
  stop(paste("Год", target_year, "не найден. Доступны:", paste(head(years, 10), collapse = ", ")))
}

# Фильтрация и подготовка матрицы "страны × индикаторы"
data_filtered <- data_wide[`Indicator Name` %in% found_indicators,
                           c("Country Name", "Indicator Name", target_year), with = FALSE]
setnames(data_filtered, target_year, "Value")
data_filtered <- na.omit(data_filtered)

# Преобразование в широкий формат
data_wide_format <- dcast(data_filtered, `Country Name` ~ `Indicator Name`, value.var = "Value")
rownames(data_wide_format) <- data_wide_format$`Country Name`
my_data <- data_wide_format[, -1, drop = FALSE]
my_data <- na.omit(my_data)

if (nrow(my_data) < 3) stop("Слишком мало стран для анализа. Измените год или индикаторы.")
cat("Размер матрицы:", dim(my_data), "\n")
cat("Используемые индикаторы:", colnames(my_data), "\n")

# 3. Кластеризация (k-means, 4 кластера) ---------------------------------
set.seed(123)
df.stand <- scale(my_data)
wss <- sapply(1:10, function(k) kmeans(df.stand, k, nstart = 10)$tot.withinss)
# plot(1:10, wss, type = "b")  # для выбора числа кластеров
clus <- kmeans(df.stand, centers = 4)
clusters <- clus$cluster
cat("Размеры кластеров:\n")
print(table(clusters))

# 4. Базовые статистики по признакам -------------------------------------
stats <- t(apply(my_data, 2, function(x) {
  c(Минимум = min(x),
    Максимум = max(x),
    Среднее = mean(x),
    Отклонение = sd(x),
    Корреляция = cor(x, clusters))
}))
print(round(stats, 3))

# 5. Поиск структуры вслепую (PCA) ---------------------------------------
Y <- as.data.frame(my_data)
mod.pca <- rda(Y ~ 1)
summary(mod.pca, scale = TRUE, scaling = 2)

# Извлекаем координаты сайтов
scores_pca <- scores(mod.pca, scaling = 2)
n_pc <- ncol(scores_pca$sites)  # количество главных компонент

if (n_pc >= 2) {
  pca.scores <- as.data.frame(scores_pca$sites[, 1:2])
  colnames(pca.scores) <- c("PC1", "PC2")
  pca.scores$cluster <- as.factor(clusters)

  # Выпуклые оболочки
  hull_list <- lapply(unique(pca.scores$cluster), function(c) {
    f <- subset(pca.scores, cluster == c)
    if (nrow(f) >= 3) f[chull(f[, 1:2]), ] else NULL
  })
  hull <- do.call(rbind, hull_list)

  # Доли дисперсии
  axX <- paste("PC1 (", as.integer(100 * mod.pca$CA$eig[1] / sum(mod.pca$CA$eig)), "%)")
  axY <- paste("PC2 (", as.integer(100 * mod.pca$CA$eig[2] / sum(mod.pca$CA$eig)), "%)")

  # График PCA
  p <- ggplot() +
    geom_polygon(data = hull,
                 aes(x = PC1, y = PC2, fill = cluster), alpha = 0.4, linetype = 0) +
    geom_point(data = pca.scores,
               aes(x = PC1, y = PC2, shape = cluster, colour = cluster), size = 3) +
    scale_colour_manual(values = rainbow(length(unique(clusters)))) +
    xlab(axX) + ylab(axY) + coord_equal() + theme_bw()
  print(p)
} else {
  cat("В PCA получена только одна компонента. График не построен.\n")
}

# 6. RDA с учётом кластеров ---------------------------------------------
FAC <- as.factor(clusters)
mod.rda <- rda(Y ~ FAC)
summary(mod.rda)

# Координаты в RDA
scores_rda <- scores(mod.rda, display = "sites", scaling = 3)
if (ncol(scores_rda) >= 2) {
  rda.scores <- as.data.frame(scores_rda[, 1:2])
  colnames(rda.scores) <- c("RDA1", "PC1")
  rda.scores <- cbind(FAC, rda.scores)

  centroids <- aggregate(cbind(RDA1, PC1) ~ FAC, rda.scores, mean)
  f_se <- function(z) sd(z) / sqrt(length(z))
  se <- aggregate(cbind(RDA1, PC1) ~ FAC, rda.scores, f_se)
  names(se) <- c("FAC", "RDA1.se", "PC1.se")
  centroids <- merge(centroids, se, by = "FAC")

  p2 <- ggplot() +
    geom_point(data = rda.scores,
               aes(x = RDA1, y = PC1, shape = FAC, colour = FAC), size = 2) +
    stat_ellipse(data = rda.scores, aes(x = RDA1, y = PC1, fill = FAC), type = "norm",
                 geom = "polygon", level = 0.8, alpha = 0.2) + # Added type = "norm" for robustness
    geom_point(data = centroids, aes(x = RDA1, y = PC1, colour = FAC),
               shape = 1, size = 4) +
    scale_colour_manual(values = rainbow(length(unique(FAC)))) +
    scale_fill_manual(values = rainbow(length(unique(FAC)))) +
    theme_bw()
  print(p2)
} else {
  cat("RDA выдал только одну ось. График не построен.\n")
}

# 7. Оценка достоверности различий ---------------------------------------
mod.man <- manova(as.matrix(Y) ~ FAC)
cat("\nРезультаты MANOVA:\n")
print(summary(mod.man))

# Hotelling test для двух наиболее многочисленных кластеров
tab <- table(FAC)
if (length(tab) >= 2) {
  largest <- names(sort(tab, decreasing = TRUE)[1:2])
  group1 <- Y[FAC == largest[1], ]
  group2 <- Y[FAC == largest[2], ]

  cat("\nHotelling test для кластеров", largest[1], "и", largest[2], ":\n")
  hot <- hotelling.test(group1, group2, perm = TRUE, B = 1000, progBar = FALSE)
  print(hot)
} else {
  cat("\nНедостаточно кластеров для Hotelling test.\n")
}
```
