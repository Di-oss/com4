```r

import kagglehub
import os

# Скачиваем датасет
path = kagglehub.dataset_download("theworldbank/world-development-indicators")
print("Dataset path:", path)

# Ищем WDIData.csv (основной файл с данными)
csv_files = []
for root, dirs, files in os.walk(path):
    for f in files:
        if f == "WDIData.csv":
            csv_files.append(os.path.join(root, f))

if csv_files:
    csv_path = csv_files[0]
    print("Data file found:", csv_path)
    # Сохраняем путь в файл для R
    with open('/tmp/csv_path.txt', 'w') as f:
        f.write(csv_path)
else:
    print("WDIData.csv not found. Available files:")
    for root, dirs, files in os.walk(path):
        for f in files:
            if f.endswith('.csv'):
                print(os.path.join(root, f))
```

```r
# Загрузка пакетов
library(data.table)
library(cluster)
library(reshape2)
library(ggplot2)

# Путь к CSV (из вывода Python-ячейки)
csv_path <- "/kaggle/input/world-development-indicators/wdi-csv-zip-57-mb-/WDIData.csv"
if (!file.exists(csv_path)) stop("Файл не найден")

# Чтение данных с заголовками
data_wide <- fread(csv_path, header = TRUE)

# Список желаемых индикаторов и их возможных точных названий в данных
# Используем список с альтернативными названиями для лучшего сопоставления
indicator_map <- list(
  "GDP per capita" = c("GDP per capita (current US$)"),
  "Population" = c("Population, total"),
  "Health expenditure per capita" = c("Health expenditure per capita (current US$", "Current health expenditure per capita (current US$)"),
  "School enrollment, primary" = c("School enrollment, primary (% gross)", "School enrollment, primary, gross %"),
  "CO2 emissions per capita" = c("CO2 emissions (metric tons per capita)")
)

found_indicators_for_analysis <- c()
all_available_names <- unique(data_wide$`Indicator Name`)

# Ищем лучшие совпадения для каждого индикатора
for (category in names(indicator_map)) {
  matched_name <- NULL
  for (potential_name in indicator_map[[category]]) {
    if (potential_name %in% all_available_names) {
      matched_name <- potential_name
      break # Найдено совпадение для этой категории
    }
  }
  if (!is.null(matched_name)) {
    found_indicators_for_analysis <- c(found_indicators_for_analysis, matched_name)
  } else {
    warning(paste0("Не удалось найти подходящий индикатор для категории '", category, "'. Он будет пропущен."))
  }
}

# Обновляем список индикаторов, используя найденные названия
indicators <- unique(found_indicators_for_analysis)

if (length(indicators) < 2) {
  stop("Слишком мало индикаторов найдено в данных для анализа. Проверьте ключевые слова индикаторов или источник данных.")
}

cat("Окончательный список индикаторов, выбранных для анализа:\n")
print(indicators)

# Выбираем год
target_year <- "2017"

# Проверяем, есть ли нужный год в данных
years <- names(data_wide)[!names(data_wide) %in% c("Country Name", "Country Code", "Indicator Name", "Indicator Code")]
if (!target_year %in% years) {
  stop(paste("Год", target_year, "не найден. Доступные годы:", paste(head(years, 10), collapse = ", ")))
}

# Фильтруем строки по найденным индикаторам и оставляем нужные колонки
data_filtered <- data_wide[`Indicator Name` %in% indicators, c("Country Name", "Indicator Name", target_year), with = FALSE]

# Удаляем строки, где значение для целевого года равно NA
data_filtered <- na.omit(data_filtered)

# Проверяем, какие индикаторы остались после удаления NA
valid_indicators_after_na_filter <- unique(data_filtered$`Indicator Name`)
if (length(setdiff(indicators, valid_indicators_after_na_filter)) > 0) {
  warning("Некоторые индикаторы не имели действительных данных для целевого года ни для одной страны и были исключены: ", paste(setdiff(indicators, valid_indicators_after_na_filter), collapse = ", "))
  indicators <- valid_indicators_after_na_filter # Обновляем список индикаторов
}

if (length(indicators) < 2) {
  stop("После фильтрации осталось слишком мало индикаторов с полными данными для целевого года. Измените индикаторы или год.")
}

# Переименовываем колонку с годом в "Value" для dcast
setnames(data_filtered, target_year, "Value")

# Разворачиваем в широкий формат: строки – страны, колонки – индикаторы
data_indicators_wide <- dcast(data_filtered, `Country Name` ~ `Indicator Name`, value.var = "Value")

# Удаляем страны, у которых есть пропуски хотя бы по одному из индикаторов (после dcast)
original_country_count <- nrow(data_indicators_wide)
data_indicators_wide_clean <- na.omit(data_indicators_wide)

if (nrow(data_indicators_wide_clean) < original_country_count) {
  cat(paste0("Удалено ", original_country_count - nrow(data_indicators_wide_clean), " стран из-за отсутствия данных по некоторым индикаторам.\n"))
}

# Проверяем, достаточно ли стран осталось для кластеризации
if (nrow(data_indicators_wide_clean) < 3) {
  stop("Слишком мало стран с полными данными по всем выбранным индикаторам для кластеризации. Попробуйте другой год или измените список индикаторов.")
}

# Устанавливаем названия строк и удаляем колонку со странами
rownames(data_indicators_wide_clean) <- data_indicators_wide_clean$`Country Name`
my_data <- data_indicators_wide_clean[, -1, drop = FALSE]

cat("Размер матрицы после обработки:", dim(my_data), "\n")

# ------------------- КЛАСТЕРИЗАЦИЯ -------------------
# Тепловая карта (формируем «дендрограмму»)
x <- as.matrix(my_data)
rc <- rainbow(nrow(x), start = 0, end = .3)
cc <- rainbow(ncol(x), start = 0, end = .3)
heatmap(x, scale = "col", RowSideColors = rc, ColSideColors = cc,
        margins = c(10,10), cexCol = 1.5, cexRow = 1)

# Оптимальное число кластеров
df.stand <- scale(my_data)
wss <- sapply(1:10, function(k) kmeans(df.stand, k, nstart = 10)$tot.withinss)
plot(1:10, wss, type = "b", pch = 19, xlab = "Число кластеров K",
     ylab = "Общая внутригрупповая сумма квадратов")

# K-means с 4 кластерами (можно изменить)
set.seed(123)
clus <- kmeans(df.stand, centers = 4)
print(clus)

# Таблица расстояний между кластерами
n <- nrow(df.stand)
euc.dist <- as.matrix(dist(df.stand))
dist_melt <- melt(euc.dist)
df.stand <- as.data.frame(df.stand)
df.stand$cluster <- clus$cluster
pairs <- data.frame(dist = dist_melt,
                    ca = as.vector(outer(1:n, 1:n, function(a,b) df.stand$cluster[a])),
                    cb = as.vector(outer(1:n, 1:n, function(a,b) df.stand$cluster[b])))
dist_between <- dcast(pairs, ca ~ cb, value.var = "dist.value", mean)
print(dist_between)

# PCA и визуализация
pca <- prcomp(my_data, center = TRUE, scale = TRUE)
d <- data.frame(x = pca$x[,1], y = pca$x[,2], cluster = clus$cluster)

h <- do.call(rbind, lapply(unique(d$cluster), function(c) {
  f <- subset(d, cluster == c)
  f[chull(f), ]
}))

ggplot(d, aes(x, y, color = as.factor(cluster), label = cluster)) +
  geom_text(size = 3) +
  geom_polygon(data = h, aes(group = cluster), alpha = 0.4, linetype = 0) +
  theme(legend.position = "none") +
  labs(title = "Кластеризация стран по экономическим показателям")
```
