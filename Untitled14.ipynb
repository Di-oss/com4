```python
# Модуль 1: Товары и категории
# Интернет-магазин "Все и сразу"

from datetime import datetime
import uuid

class Category:
    """Класс категории товаров"""
    def __init__(self, name, description="", parent_id=None):
        self.id = str(uuid.uuid4())[:8]
        self.name = name
        self.description = description
        self.parent_id = parent_id
        self.created_at = datetime.now()
        self.products = []  # Связь с товарами

    def add_product(self, product):
        self.products.append(product)
        print(f"Товар {product.name} добавлен в категорию {self.name}")

    def get_product_count(self):
        return len(self.products)

    def __str__(self):
        return f"{self.name} (ID: {self.id}, товаров: {len(self.products)})"

class Product:
    """Класс товара"""
    def __init__(self, name, price, quantity, category_id=None, brand="", description=""):
        self.id = str(uuid.uuid4())[:8]
        self.name = name
        self.price = price
        self.quantity_in_stock = quantity
        self.category_id = category_id
        self.brand = brand
        self.description = description
        self.created_at = datetime.now()
        self.updated_at = datetime.now()
        self.views_count = 0
        self.discount_percent = 0
        self.reviews = []  # Связь с отзывами

    def apply_discount(self, percent):
        if 0 <= percent <= 100:
            self.discount_percent = percent
            self.updated_at = datetime.now()
            print(f"Скидка {percent}% применена к товару {self.name}")
            return True
        print("Ошибка: скидка от 0 до 100%")
        return False

    def get_final_price(self):
        if self.discount_percent > 0:
            return self.price * (1 - self.discount_percent / 100)
        return self.price

    def check_availability(self, requested_quantity):
        return self.quantity_in_stock >= requested_quantity

    def reduce_stock(self, quantity):
        if self.check_availability(quantity):
            self.quantity_in_stock -= quantity
            self.updated_at = datetime.now()
            return True
        return False

    def add_stock(self, quantity):
        self.quantity_in_stock += quantity
        self.updated_at = datetime.now()
        print(f"Добавлено {quantity} ед. Остаток: {self.quantity_in_stock}")

    def increment_views(self):
        self.views_count += 1

    def get_info(self):
        info = f"\n=== ТОВАР: {self.name} ===\n"
        info += f"ID: {self.id}\n"
        info += f"Бренд: {self.brand}\n"
        info += f"Цена: {self.price:.2f} руб.\n"
        if self.discount_percent > 0:
            info += f"Скидка: {self.discount_percent}%\n"
            info += f"Цена со скидкой: {self.get_final_price():.2f} руб.\n"
        info += f"В наличии: {self.quantity_in_stock} шт.\n"
        info += f"Просмотров: {self.views_count}\n"
        info += f"Описание: {self.description}\n"
        info += "="*30
        return info

    def __str__(self):
        price_str = f"{self.get_final_price():.2f} руб."
        if self.discount_percent > 0:
            price_str += f" (-{self.discount_percent}%)"
        return f"{self.name} | {self.brand} | {price_str} | {self.quantity_in_stock} шт."

class CatalogManager:
    """Менеджер каталога"""
    def __init__(self):
        self.categories = {}  # id -> Category
        self.products = {}    # id -> Product
        self.products_by_category = {}  # category_id -> [product_ids]

    def add_category(self, name, description="", parent_id=None):
        category = Category(name, description, parent_id)
        self.categories[category.id] = category
        print(f"Категория '{name}' создана (ID: {category.id})")
        return category

    def add_product(self, name, price, quantity, category_id=None, brand="", description=""):
        # Проверка существования категории
        if category_id and category_id not in self.categories:
            print("Ошибка: категория не найдена")
            return None

        product = Product(name, price, quantity, category_id, brand, description)
        self.products[product.id] = product

        # Связь с категорией
        if category_id:
            if category_id not in self.products_by_category:
                self.products_by_category[category_id] = []
            self.products_by_category[category_id].append(product.id)
            self.categories[category_id].add_product(product)

        print(f"Товар '{name}' создан (ID: {product.id})")
        return product

    def get_category(self, category_id):
        return self.categories.get(category_id)

    def get_product(self, product_id):
        return self.products.get(product_id)

    def get_products_by_category(self, category_id):
        product_ids = self.products_by_category.get(category_id, [])
        return [self.products[pid] for pid in product_ids if pid in self.products]

    def search_products(self, query):
        query = query.lower()
        results = []
        for product in self.products.values():
            if (query in product.name.lower() or
                query in product.brand.lower() or
                query in product.description.lower()):
                results.append(product)
                product.increment_views()
        return results

    def filter_by_price(self, min_price=0, max_price=float('inf')):
        results = []
        for product in self.products.values():
            price = product.get_final_price()
            if min_price <= price <= max_price:
                results.append(product)
        return results

    def filter_by_brand(self, brand):
        brand = brand.lower()
        return [p for p in self.products.values() if p.brand.lower() == brand]

    def get_low_stock_products(self, threshold=5):
        return [p for p in self.products.values() if p.quantity_in_stock < threshold]

    def get_most_viewed(self, limit=5):
        sorted_products = sorted(self.products.values(),
                                key=lambda p: p.views_count, reverse=True)
        return sorted_products[:limit]

    def update_product(self, product_id, **kwargs):
        product = self.products.get(product_id)
        if not product:
            print("Товар не найден")
            return False

        for key, value in kwargs.items():
            if hasattr(product, key):
                setattr(product, key, value)

        product.updated_at = datetime.now()
        print(f"Товар {product.name} обновлен")
        return True

    def delete_product(self, product_id):
        if product_id in self.products:
            product = self.products[product_id]
            # Удаляем из связей с категорией
            if product.category_id and product.category_id in self.products_by_category:
                if product_id in self.products_by_category[product.category_id]:
                    self.products_by_category[product.category_id].remove(product_id)

            del self.products[product_id]
            print(f"Товар {product.name} удален")
            return True
        return False

    def list_categories(self):
        if not self.categories:
            print("Нет категорий")
            return

        print("\n=== КАТЕГОРИИ ===")
        for category in self.categories.values():
            print(f"{category.id} | {category.name} - {category.get_product_count()} товаров")
        print("="*30)

    def list_products(self, products=None):
        if products is None:
            products = self.products.values()

        if not products:
            print("Нет товаров")
            return

        print("\n=== ТОВАРЫ ===")
        for i, product in enumerate(products, 1):
            print(f"{i}. {product}")
        print("="*30)

# Тестовые данные
catalog = CatalogManager()

# Добавляем тестовые категории
electronics = catalog.add_category("Электроника", "Бытовая техника и гаджеты")
phones = catalog.add_category("Смартфоны", "Мобильные телефоны", electronics.id)
laptops = catalog.add_category("Ноутбуки", "Портативные компьютеры", electronics.id)
clothing = catalog.add_category("Одежда", "Модная одежда и аксессуары")

# Добавляем тестовые товары
catalog.add_product(
    name="iPhone 15 Pro",
    price=99990.00,
    quantity=15,
    category_id=phones.id,
    brand="Apple",
    description="Флагманский смартфон Apple"
)

catalog.add_product(
    name="Samsung Galaxy S24",
    price=89990.00,
    quantity=8,
    category_id=phones.id,
    brand="Samsung",
    description="Флагманский смартфон Samsung"
)

catalog.add_product(
    name="MacBook Pro 14",
    price=199990.00,
    quantity=5,
    category_id=laptops.id,
    brand="Apple",
    description="Профессиональный ноутбук"
)

catalog.add_product(
    name="Футболка хлопковая",
    price=1990.00,
    quantity=50,
    category_id=clothing.id,
    brand="Nike",
    description="Базовая футболка"
)

# Главный цикл
while True:
    print("\n" + "="*50)
    print("МОДУЛЬ ТОВАРОВ И КАТЕГОРИЙ")
    print("="*50)
    print("1 - Управление категориями")
    print("2 - Управление товарами")
    print("3 - Поиск и фильтры")
    print("4 - Просмотр каталога")
    print("5 - Работа с товаром")
    print("0 - Выход")

    choice = input("\nВыберите действие: ")

    if choice == "1":
        # Управление категориями
        while True:
            print("\n--- КАТЕГОРИИ ---")
            print("1 - Добавить категорию")
            print("2 - Список категорий")
            print("3 - Товары категории")
            print("0 - Назад")

            sub = input("Выберите: ")

            if sub == "1":
                name = input("Название категории: ")
                desc = input("Описание (Enter - пропустить): ")

                # Выбор родительской категории
                parent_id = None
                if catalog.categories:
                    print("Доступные родительские категории:")
                    for cat in catalog.categories.values():
                        print(f"  {cat.id} - {cat.name}")
                    parent = input("ID родительской категории (Enter - без родителя): ")
                    if parent in catalog.categories:
                        parent_id = parent

                catalog.add_category(name, desc, parent_id)

            elif sub == "2":
                catalog.list_categories()

            elif sub == "3":
                catalog.list_categories()
                cat_id = input("Введите ID категории: ")
                products = catalog.get_products_by_category(cat_id)
                if products:
                    print(f"\nТовары в категории {cat_id}:")
                    catalog.list_products(products)
                else:
                    print("В этой категории нет товаров")

            elif sub == "0":
                break

    elif choice == "2":
        # Управление товарами
        while True:
            print("\n--- ТОВАРЫ ---")
            print("1 - Добавить товар")
            print("2 - Удалить товар")
            print("3 - Обновить товар")
            print("4 - Применить скидку")
            print("5 - Пополнить склад")
            print("0 - Назад")

            sub = input("Выберите: ")

            if sub == "1":
                name = input("Название товара: ")
                try:
                    price = float(input("Цена: "))
                    quantity = int(input("Количество на складе: "))
                except ValueError:
                    print("Ошибка ввода")
                    continue

                brand = input("Бренд: ")
                desc = input("Описание (Enter - пропустить): ")

                # Выбор категории
                category_id = None
                if catalog.categories:
                    print("Доступные категории:")
                    for cat in catalog.categories.values():
                        print(f"  {cat.id} - {cat.name}")
                    cat = input("ID категории (Enter - без категории): ")
                    if cat in catalog.categories:
                        category_id = cat

                catalog.add_product(name, price, quantity, category_id, brand, desc)

            elif sub == "2":
                catalog.list_products()
                prod_id = input("Введите ID товара для удаления: ")
                catalog.delete_product(prod_id)

            elif sub == "3":
                catalog.list_products()
                prod_id = input("Введите ID товара для обновления: ")
                product = catalog.get_product(prod_id)
                if product:
                    print("Оставьте поле пустым, чтобы не менять")
                    new_name = input(f"Название [{product.name}]: ")
                    if new_name:
                        product.name = new_name

                    try:
                        new_price = input(f"Цена [{product.price}]: ")
                        if new_price:
                            product.price = float(new_price)

                        new_brand = input(f"Бренд [{product.brand}]: ")
                        if new_brand:
                            product.brand = new_brand
                    except ValueError:
                        print("Ошибка ввода")

                    product.updated_at = datetime.now()
                    print("Товар обновлен")

            elif sub == "4":
                catalog.list_products()
                prod_id = input("Введите ID товара: ")
                product = catalog.get_product(prod_id)
                if product:
                    try:
                        discount = float(input("Размер скидки (%): "))
                        product.apply_discount(discount)
                    except ValueError:
                        print("Ошибка ввода")

            elif sub == "5":
                catalog.list_products()
                prod_id = input("Введите ID товара: ")
                product = catalog.get_product(prod_id)
                if product:
                    try:
                        qty = int(input("Количество для добавления: "))
                        product.add_stock(qty)
                    except ValueError:
                        print("Ошибка ввода")

            elif sub == "0":
                break

    elif choice == "3":
        # Поиск и фильтры
        while True:
            print("\n--- ПОИСК И ФИЛЬТРЫ ---")
            print("1 - Поиск по названию/бренду")
            print("2 - Фильтр по цене")
            print("3 - Фильтр по бренду")
            print("4 - Товары с низким остатком")
            print("5 - Популярные товары")
            print("0 - Назад")

            sub = input("Выберите: ")

            if sub == "1":
                query = input("Поисковый запрос: ")
                results = catalog.search_products(query)
                if results:
                    print(f"\nНайдено товаров: {len(results)}")
                    catalog.list_products(results)
                else:
                    print("Ничего не найдено")

            elif sub == "2":
                try:
                    min_p = float(input("Минимальная цена (Enter - 0): ") or 0)
                    max_p = input("Максимальная цена (Enter - без ограничений): ")
                    max_p = float(max_p) if max_p else float('inf')
                    results = catalog.filter_by_price(min_p, max_p)
                    catalog.list_products(results)
                except ValueError:
                    print("Ошибка ввода")

            elif sub == "3":
                brand = input("Бренд: ")
                results = catalog.filter_by_brand(brand)
                if results:
                    catalog.list_products(results)
                else:
                    print(f"Товаров бренда {brand} не найдено")

            elif sub == "4":
                try:
                    threshold = int(input("Порог остатка (Enter - 5): ") or 5)
                    results = catalog.get_low_stock_products(threshold)
                    if results:
                        print(f"\nТовары с остатком менее {threshold}:")
                        catalog.list_products(results)
                    else:
                        print("Все товары в достаточном количестве")
                except ValueError:
                    print("Ошибка ввода")

            elif sub == "5":
                try:
                    limit = int(input("Сколько показать (Enter - 5): ") or 5)
                    results = catalog.get_most_viewed(limit)
                    print("\nСамые популярные товары:")
                    catalog.list_products(results)
                except ValueError:
                    print("Ошибка ввода")

            elif sub == "0":
                break

    elif choice == "4":
        # Просмотр каталога
        while True:
            print("\n--- ПРОСМОТР КАТАЛОГА ---")
            print("1 - Все категории")
            print("2 - Все товары")
            print("3 - Товары по категориям")
            print("0 - Назад")

            sub = input("Выберите: ")

            if sub == "1":
                catalog.list_categories()

            elif sub == "2":
                catalog.list_products()

            elif sub == "3":
                catalog.list_categories()
                cat_id = input("Введите ID категории: ")
                products = catalog.get_products_by_category(cat_id)
                if products:
                    category = catalog.get_category(cat_id)
                    print(f"\n--- {category.name} ---")
                    catalog.list_products(products)
                else:
                    print("В этой категории нет товаров")

            elif sub == "0":
                break

    elif choice == "5":
        # Работа с конкретным товаром
        prod_id = input("Введите ID товара: ")
        product = catalog.get_product(prod_id)
        if not product:
            print("Товар не найден")
            continue

        while True:
            print("\n" + "-"*30)
            print(f"ТОВАР: {product.name}")
            print("-"*30)
            print("1 - Показать информацию")
            print("2 - Применить скидку")
            print("3 - Изменить цену")
            print("4 - Изменить количество")
            print("5 - Добавить описание")
            print("0 - Назад")

            sub = input("Выберите: ")

            if sub == "1":
                print(product.get_info())

            elif sub == "2":
                try:
                    disc = float(input("Скидка (%): "))
                    product.apply_discount(disc)
                except ValueError:
                    print("Ошибка ввода")

            elif sub == "3":
                try:
                    new_price = float(input("Новая цена: "))
                    product.price = new_price
                    product.updated_at = datetime.now()
                    print("Цена изменена")
                except ValueError:
                    print("Ошибка ввода")

            elif sub == "4":
                try:
                    qty = int(input("Новое количество: "))
                    product.quantity_in_stock = qty
                    product.updated_at = datetime.now()
                    print("Количество изменено")
                except ValueError:
                    print("Ошибка ввода")

            elif sub == "5":
                desc = input("Описание: ")
                product.description = desc
                product.updated_at = datetime.now()
                print("Описание добавлено")

            elif sub == "0":
                break

    elif choice == "0":
        print("Программа завершена")
        break

    else:
        print("Неверный выбор. Попробуйте снова.")
```
